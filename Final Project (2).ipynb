Pair plots: Show relationships between multiple variables at once.

### 8.2.3 Outlier Detection
Here we get 3 columns should have the unusual and extreme values.So we must ensure that detect them by not affect the entire data 

1.Age column-Here check the compplete records of patients who reqiured to be an outlier.


```python
df[df['Age'] == 120]
```

Here we get 5 records they all are in different diseases and departments.So it seems to be a potential outlier.So we can't delete the recods and we have to detect it using quartile method.First we simply draw a box plot for seeing the outliers.

```python
sns.boxplot(x=df['Age'])
plt.title('Age Outlier Detection')
plt.show()
```

The points outside the whiskers are potential outliers.
so we are going to use the IQR method.

```python
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
```

```python
age_outliers = df[
    (df['Age'] < lower_limit) |
    (df['Age'] > upper_limit)
]
age_outliers[['Patient_ID', 'Age', 'Gender', 'Disease', 'Severity']]
```

```python
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
```

```python
df['Age'].max()
```

Five records with an age of 120 years were identified. Although age 120 is an extremely rare value, the IQR method gave an upper limit of 122 years. Therefore, these records were not classified as statistical outliers and were retained for analysis.

According to our data set we can't remove the rows because outlier detection is depend on the scenarios. So here diseases are happened with the aged people.So we have to keep it.

2.Length of stay column

```python
sns.boxplot(x=df['Length_of_Stay'])
plt.title('Box Plot of Length of Stay')
plt.show()
```

```python
Q1 = df['Length_of_Stay'].quantile(0.25)
Q3 = df['Length_of_Stay'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)

los_outliers = df[
    (df['Length_of_Stay'] < lower_limit) |
    (df['Length_of_Stay'] > upper_limit)
]

len(los_outliers)
```

The IQR method identified 256 potential outliers in the Length of Stay column. These values represent unusually long hospital stays and were investigated. Since long stays can be valid in healthcare data, the values were retained and not removed.
